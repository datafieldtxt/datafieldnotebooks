## Lists

Lists are Python's most versatile data structure — ordered, mutable sequences that can hold elements of any type, including other lists. They are created with square brackets `[]`. Lists support indexing and slicing (just like strings), and they have a rich set of methods for adding, removing, and rearranging elements. Lists are the right choice when order matters and you need to add or remove items.

```python
# List creation and basic operations
temperatures = [22.1, 19.8, 25.4, 18.3, 27.6, 23.0, 20.5]

print(f"Length: {len(temperatures)}")
print(f"First:  {temperatures[0]}")
print(f"Last:   {temperatures[-1]}")
print(f"Slice:  {temperatures[2:5]}")
print(f"Max:    {max(temperatures)}")
print(f"Min:    {min(temperatures)}")
print(f"Sum:    {sum(temperatures):.1f}")
print(f"Avg:    {sum(temperatures)/len(temperatures):.2f}")

# Lists can hold mixed types (though homogeneous lists are more common)
mixed = [42, "hello", 3.14, True, None, [1, 2, 3]]
print(f"\nMixed list: {mixed}")
```

```python
# List methods — modifying lists in place
task_list = ["write tests", "refactor code", "update docs"]

# append() — add one item to the end
task_list.append("deploy to staging")
print("After append:", task_list)

# insert() — add at a specific index
task_list.insert(0, "fix critical bug")  # Add at front
print("After insert:", task_list)

# extend() — add all items from another iterable
more_tasks = ["write release notes", "notify team"]
task_list.extend(more_tasks)
print(f"After extend: {len(task_list)} tasks")

# remove() — remove first occurrence by value
task_list.remove("update docs")
print("After remove:", task_list)

# pop() — remove and return item by index (default: last)
last = task_list.pop()
print(f"Popped: '{last}'")
first = task_list.pop(0)
print(f"Popped first: '{first}'")

# sort() — sort in place; sorted() returns a new list
numbers = [64, 34, 25, 12, 22, 11, 90]
numbers.sort()
print("Sorted:", numbers)
numbers.sort(reverse=True)
print("Reversed:", numbers)
```

## Tuples

Tuples are like lists but **immutable** — once created, their contents cannot be changed. They are created with parentheses `()` or just commas. Tuples are ideal for data that should not be modified: coordinates, RGB colours, database records, function return values. Tuple **unpacking** is one of Python's most elegant features, allowing you to assign multiple variables at once from a single tuple.

```python
# Tuple creation
point = (3.0, 4.0)  # 2D coordinate
colour_rgb = (255, 128, 0)  # Orange in RGB
record = ("alice", 28, "engineer", True)  # Database-like record

# Tuples support indexing and slicing (like lists)
print(point[0])         # 3.0
print(colour_rgb[-1])   # 0
print(record[1:3])      # (28, 'engineer')

# Immutability — this raises a TypeError
try:
    point[0] = 5.0
except TypeError as e:
    print(f"Cannot modify tuple: {e}")

# Creating a tuple with one element requires a trailing comma
single = (42,)  # This is a tuple
not_tuple = (42)  # This is just the integer 42 in parentheses!
print(f"single: {type(single)}, not_tuple: {type(not_tuple)}")
```

```python
# Tuple unpacking — assign multiple variables from a tuple
name, age, job, active = record
print(f"{name} is a {age}-year-old {job}")

# Swap variables without a temp variable (classic Python trick)
a, b = 10, 20
print(f"Before: a={a}, b={b}")
a, b = b, a  # Tuple packing and unpacking in one line
print(f"After swap: a={a}, b={b}")

# Extended unpacking with *
first, *middle, last = [10, 20, 30, 40, 50]
print(f"first={first}, middle={middle}, last={last}")
# first=10, middle=[20, 30, 40], last=50

# Practical: unpack function return values
def get_dimensions():
    return 1920, 1080  # Python packs these as a tuple

width, height = get_dimensions()
print(f"Resolution: {width}x{height}")
print(f"Aspect ratio: {width/height:.4f}")
```

## Sets

Sets are unordered collections of **unique** elements. Duplicate values are automatically removed. Sets are ideal for membership testing (very fast — O(1) average), removing duplicates from a sequence, and performing mathematical set operations like union, intersection, and difference. Sets are mutable (you can add/remove elements), but the elements themselves must be immutable (hashable).

```python
# Set creation — duplicates are automatically removed
tags_post1 = {"python", "beginner", "tutorial", "python"}  # 'python' appears twice
print(tags_post1)  # {'python', 'beginner', 'tutorial'} — no duplicates
print(f"Type: {type(tags_post1)}")

# Create a set from a list (common way to deduplicate)
rawscores = [85, 92, 78, 85, 91, 78, 92, 88]
unique_scores = set(rawscores)
print(f"Unique scores: {sorted(unique_scores)}")  # Sort for display

# Membership testing — very fast O(1) average
permitted_extensions = {".jpg", ".jpeg", ".png", ".gif", ".webp"}
filename = "profile_photo.jpg"
extension = "." + filename.rsplit(".", 1)[-1]
if extension in permitted_extensions:
    print(f"{filename} is allowed")
else:
    print(f"{filename} is not allowed")
```

```python
# Set operations — like Venn diagrams
python_developers = {"Alice", "Bob", "Carol", "David", "Eve"}
java_developers   = {"Bob", "David", "Frank", "Grace"}

# Union |  — everyone who knows at least one language
all_devs = python_developers | java_developers
print(f"All developers: {sorted(all_devs)}")

# Intersection &  — those who know BOTH languages
bilingual = python_developers & java_developers
print(f"Know both: {sorted(bilingual)}")

# Difference -  — Python devs who do NOT know Java
python_only = python_developers - java_developers
print(f"Python only: {sorted(python_only)}")

# Symmetric difference ^  — those who know exactly one language
exclusive = python_developers ^ java_developers
print(f"Exactly one language: {sorted(exclusive)}")

# Subset and superset checks
print(f"bilingual is subset of python: {bilingual <= python_developers}")
print(f"python is superset of bilingual: {python_developers >= bilingual}")
```

## Dictionaries

Dictionaries store **key-value pairs**, where keys must be unique and immutable (strings, numbers, tuples). Values can be anything. Dictionaries are Python's most important data structure after lists — they are used to represent objects, configuration settings, API responses, and more. Since Python 3.7, dictionaries maintain **insertion order**. Looking up a value by key is very fast (O(1) average).

```python
# Dictionary creation
student = {
    "id": 10042,
    "name": "Maria Santos",
    "age": 21,
    "grades": {"math": 91, "physics": 87, "english": 95},  # Nested dict
    "courses": ["CS101", "MATH201", "PHYS101"]            # List as value
}

# Accessing values
print(student["name"])                    # Maria Santos
print(student["grades"]["physics"])       # 87 — nested access

# get() — safe access with default value
print(student.get("gpa", "Not calculated"))  # Not calculated (key doesn't exist)
print(student.get("age", 0))                 # 21 (key exists)

# Adding and updating
student["email"] = "maria@university.edu"  # Add new key
student["age"] = 22                         # Update existing key

# Deleting
del student["id"]                           # Remove a key
popped = student.pop("age")                 # Remove and return
print(f"Removed age: {popped}")
```

```python
# Dictionary methods — keys(), values(), items(), update()
inventory = {
    "apple":   {"qty": 150, "price": 0.99},
    "banana":  {"qty": 80,  "price": 0.59},
    "cherry":  {"qty": 200, "price": 4.99},
    "date":    {"qty": 30,  "price": 7.99},
}

# Iterating
print("Inventory report:")
for fruit, info in inventory.items():
    total = info["qty"] * info["price"]
    print(f"  {fruit:<10} qty: {info['qty']:>4}  value: ${total:>8.2f}")

# Total value calculation
total_value = sum(info["qty"] * info["price"] for info in inventory.values())
print(f"\nTotal inventory value: ${total_value:.2f}")

# update() — merge another dict (or keyword args) into this one
price_update = {"apple": {"qty": 150, "price": 1.09}, "fig": {"qty": 40, "price": 3.49}}
inventory.update(price_update)
print(f"\nApple price updated: ${inventory['apple']['price']}")
print(f"Figs added: {inventory.get('fig')}")
```

## List and Dict Comprehensions

Comprehensions are a concise, readable way to create lists, dictionaries, and sets from existing iterables. They replace many common `for` loops that build up a new collection. The syntax `[expression for item in iterable if condition]` is faster than an equivalent loop and is considered idiomatic Python. Use them when the logic fits on one line and remains readable.

```python
# List comprehensions
numbers = range(1, 21)

# Basic: square each number
squares = [n ** 2 for n in numbers]
print("Squares:", squares)

# With condition: only even squares
even_squares = [n ** 2 for n in numbers if n % 2 == 0]
print("Even squares:", even_squares)

# Transform strings
raw_names = [" Alice ", "BOB", "carol ", "  DAVID  "]
clean_names = [name.strip().title() for name in raw_names]
print("Clean names:", clean_names)  # ['Alice', 'Bob', 'Carol', 'David']

# Practical: filter and transform a dataset
products = [
    {"name": "Keyboard", "price": 79.99,  "in_stock": True},
    {"name": "Monitor",  "price": 349.00, "in_stock": False},
    {"name": "Mouse",    "price": 29.99,  "in_stock": True},
    {"name": "Webcam",   "price": 89.99,  "in_stock": True},
    {"name": "Headset",  "price": 149.99, "in_stock": False},
]

# Get names of in-stock items under $100
affordable_instock = [p["name"] for p in products if p["in_stock"] and p["price"] < 100]
print("Affordable and in stock:", affordable_instock)
```

```python
# Dict comprehensions
# Basic: create a dict mapping word to its length
words = ["Python", "is", "awesome", "and", "powerful"]
word_lengths = {word: len(word) for word in words}
print("Word lengths:", word_lengths)

# With condition: only words longer than 3 chars
long_words = {w: len(w) for w in words if len(w) > 3}
print("Long words:", long_words)

# Invert a dictionary (swap keys and values)
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in original.items()}
print("Inverted:", inverted)  # {1: 'a', 2: 'b', 3: 'c'}

# Set comprehension (uses {})
sentence = "the quick brown fox jumps over the lazy dog"
unique_lengths = {len(word) for word in sentence.split()}
print("Unique word lengths:", sorted(unique_lengths))  # [2, 3, 4, 5]
```

## Nested Data Structures

Real-world data is rarely flat. JSON APIs, databases, and configuration files all use nested structures — lists of dicts, dicts of lists, dicts of dicts, and so on. Working with nested structures requires chaining `[]` accesses or using `.get()` chains to safely navigate without errors. The `defaultdict` from `collections` is a useful tool for building nested structures incrementally.

```python
from collections import defaultdict

# A realistic nested structure: a class roster
classroom = {
    "course": "Introduction to Python",
    "instructor": "Dr. Chen",
    "students": [
        {"id": 1001, "name": "Alice",  "grades": [88, 92, 79, 95]},
        {"id": 1002, "name": "Bob",    "grades": [75, 68, 82, 71]},
        {"id": 1003, "name": "Carol",  "grades": [95, 98, 92, 97]},
        {"id": 1004, "name": "David",  "grades": [60, 55, 70, 65]},
    ]
}

# Access nested values
print(f"Course: {classroom['course']}")
print(f"Instructor: {classroom['instructor']}")
print(f"Students enrolled: {len(classroom['students'])}")

# Compute each student's average using list comprehension
print("\nStudent Averages:")
for student in classroom["students"]:
    avg = sum(student["grades"]) / len(student["grades"])
    print(f"  {student['name']:<10} avg: {avg:.1f}")

# Build a summary using defaultdict — groups students by grade category
grade_groups = defaultdict(list)  # Default value is an empty list
for student in classroom["students"]:
    avg = sum(student["grades"]) / len(student["grades"])
    category = "A" if avg >= 90 else "B" if avg >= 80 else "C" if avg >= 70 else "D"
    grade_groups[category].append(student["name"])

print("\nGrade groups:")
for category, names in sorted(grade_groups.items()):
    print(f"  {category}: {names}")
```
