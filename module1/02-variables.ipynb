Python is **dynamically typed**, which means you do not declare types explicitly — Python infers the type from the value you assign.

## Variable Naming Rules

Python variable names must follow a few rules: they can contain letters, digits, and underscores, but they cannot start with a digit. They are case-sensitive (`age` and `Age` are different variables). Python convention (PEP 8) recommends **snake_case** for variable names — all lowercase with underscores separating words. Choosing clear, descriptive names makes your code much easier to read and maintain.

```python
# Valid variable names — follow these conventions
student_name = "Carlos"
max_retries = 3
is_active = True
page_count = 250
_private_note = "internal use"  # Underscore prefix signals 'private by convention'

# Python is case-sensitive
temperature = 22.5
Temperature = 100.0  # This is a completely different variable!

print(temperature)    # 22.5
print(Temperature)    # 100.0

# Multiple assignment on one line
x, y, z = 10, 20, 30
print(x, y, z)  # 10 20 30

# Assign the same value to multiple variables
a = b = c = 0
print(a, b, c)  # 0 0 0
```

```python
# INVALID names — these would cause a SyntaxError
# 2fast = True        # Cannot start with a digit
# my-variable = 5     # Hyphens are not allowed
# class = "Math"      # 'class' is a reserved keyword

# You can see all reserved keywords like this:
import keyword
print("Python reserved keywords:")
print(keyword.kwlist)
```

## Integers (`int`)

Integers are whole numbers — positive, negative, or zero — with no decimal point. In Python 3, integers can be arbitrarily large; there is no overflow. You can use underscores in large numbers to improve readability (e.g., `1_000_000` instead of `1000000`). Integers support all arithmetic operations and are commonly used for counting, indexing, and loop control.

```python
# Integer examples
student_count = 32
negative_temp = -15
world_population = 8_100_000_000  # Underscores improve readability

print(type(student_count))       # <class 'int'>
print(type(world_population))    # <class 'int'>

# Python integers have unlimited precision
big_number = 2 ** 100
print("2^100 =", big_number)     # 1267650600228229401496703205376

# Integer division vs true division
print(7 // 2)   # 3  (floor division — always rounds toward negative infinity)
print(-7 // 2)  # -4 (not -3! floors toward negative infinity)
print(7 % 2)    # 1  (remainder)
```

## Floats (`float`)

Floats represent real numbers with a decimal point. They are stored using IEEE 754 double-precision format, which means they have about 15-17 significant digits of precision. This precision limit can cause surprising results in arithmetic (like `0.1 + 0.2` not being exactly `0.3`). For financial calculations requiring exact decimal arithmetic, Python's `decimal` module is the right tool.

```python
# Float examples
item_price = 29.99
pi_approx = 3.14159
absolute_zero = -273.15
scientific = 6.022e23   # Scientific notation: 6.022 × 10^23 (Avogadro's number)

print(type(item_price))    # <class 'float'>
print(scientific)          # 6.022e+23

# The floating-point precision issue
result = 0.1 + 0.2
print(result)              # 0.30000000000000004 — not exactly 0.3!
print(result == 0.3)       # False

# The correct way to compare floats: check if they are 'close enough'
import math
print(math.isclose(result, 0.3))   # True
print(round(result, 2) == 0.3)     # True — rounding also works for simple cases
```

## Strings (`str`)

Strings are sequences of characters used to represent text. In Python, you can create strings with single quotes, double quotes, or triple quotes (for multi-line strings). Strings are **immutable** — once created, they cannot be changed in place. Python provides a rich set of string methods for searching, transforming, and formatting text, which we will explore in depth in Lesson 7.

```python
# String examples — single, double, or triple quotes all work
city = 'Toronto'
country = "Canada"
multiline = """This string
spans multiple
lines."""

print(type(city))      # <class 'str'>
print(multiline)

# Strings can be concatenated with + and repeated with *
greeting = "Hello" + ", " + city + "!"
print(greeting)          # Hello, Toronto!

divider = "-" * 30
print(divider)           # ------------------------------

# The len() function returns the number of characters
print(len(city))         # 7
print(len(""))           # 0 — empty string has length 0
```

## Booleans (`bool`)

Booleans represent truth values: `True` or `False` (note the capital first letter — Python is case-sensitive). Booleans are the result of comparison and logical operations and control which branches of code execute in `if` statements. In Python, booleans are actually a subclass of integers: `True` equals `1` and `False` equals `0`, which allows for some clever tricks but also potential confusion.

```python
# Boolean examples
is_logged_in = True
has_permission = False

print(type(is_logged_in))   # <class 'bool'>

# Booleans result from comparisons
age = 20
print(age >= 18)    # True
print(age == 21)    # False

# Booleans are integers under the hood
print(True + True)   # 2
print(True * 5)      # 5
print(False + 10)    # 10

# Counting with booleans — a useful Python trick
scores = [85, 92, 45, 78, 55, 96, 38]
passing = [s >= 60 for s in scores]          # List of booleans
print("Passing scores:", passing)
print("Number passing:", sum(passing))        # sum() treats True as 1, False as 0
```

## The `type()` Function

The built-in `type()` function returns the type of any value or variable. This is invaluable when debugging — when your code behaves unexpectedly, checking the types of your variables often reveals the problem. For example, if you expect a number but get a string, arithmetic will fail in confusing ways. `isinstance()` is the preferred way to check types in production code because it handles inheritance correctly.

```python
# Using type() to inspect values
values = [42, 3.14, "hello", True, None, [1, 2, 3]]

for v in values:
    print(f"Value: {str(v):<15} Type: {type(v).__name__}")

# isinstance() is preferred for type checking in real code
user_input = "42"
if isinstance(user_input, str):
    print(f"\n'{user_input}' is a string — needs conversion before arithmetic")

# Checking for multiple types at once
def describe(value):
    if isinstance(value, bool):    # Check bool BEFORE int — bool is subclass of int!
        return "boolean"
    elif isinstance(value, int):
        return "integer"
    elif isinstance(value, float):
        return "float"
    else:
        return "something else"

print(describe(True))   # boolean
print(describe(42))     # integer
print(describe(3.14))   # float
```

## Type Conversion (Casting)

Sometimes you need to convert a value from one type to another. Python provides built-in functions for this: `int()`, `float()`, `str()`, and `bool()`. This is called **type casting** or **type conversion**. There are two kinds: **implicit conversion** (Python does it automatically, e.g., `int + float` gives `float`) and **explicit conversion** (you call a function to convert). Not all conversions are valid — attempting an impossible conversion raises a `ValueError`.

```python
# Implicit (automatic) type conversion
result = 5 + 2.0     # int + float -> float automatically
print(result)         # 7.0
print(type(result))   # <class 'float'>

# Explicit type conversion with int()
print(int(3.9))      # 3   — truncates (does NOT round) toward zero
print(int(-3.9))     # -3  — truncates toward zero (not floor!)
print(int("42"))     # 42  — string of digits converts fine
print(int(True))     # 1
print(int(False))    # 0

# float() conversions
print(float("3.14"))  # 3.14
print(float(7))       # 7.0
print(float("inf"))   # inf  — Python supports infinity!

# str() conversions — everything can become a string
print(str(42))        # '42'
print(str(3.14))      # '3.14'
print(str(True))      # 'True'
```

```python
# COMMON PITFALL: int() cannot convert a float string directly
# This is a very common beginner mistake!

value_str = "3.99"

# This FAILS with ValueError:
try:
    result = int(value_str)   # Can't convert "3.99" directly to int
except ValueError as e:
    print(f"Error: {e}")
    # Error: invalid literal for int() with base 10: '3.99'

# SOLUTION: Convert to float first, then to int
result = int(float(value_str))
print(f"Correct way: int(float('{value_str}')) = {result}")  # 3

# Another pitfall: bool() of non-empty strings is always True
print(bool(""))       # False — empty string is falsy
print(bool("False"))  # True  — non-empty string is ALWAYS truthy, even "False"!
print(bool("0"))      # True  — '0' is a non-empty string
print(bool(0))        # False — the integer 0 is falsy
```
