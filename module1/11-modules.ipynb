## Import Patterns

Python provides three main ways to import code. `import module` imports the whole module — you access its contents as `module.function()`. `from module import name` imports specific names directly into your namespace. `import module as alias` creates an alias — commonly used for long module names or to follow community conventions (like `import numpy as np`). Avoid `from module import *` in production code as it pollutes your namespace with unknown names.

```python
# Pattern 1: import module — access with dot notation
import math

print(f"pi:          {math.pi}")
print(f"e:           {math.e}")
print(f"sqrt(2):     {math.sqrt(2):.6f}")
print(f"log10(1000): {math.log10(1000)}")
print(f"ceil(4.1):   {math.ceil(4.1)}")
print(f"floor(4.9):  {math.floor(4.9)}")
print(f"factorial(10): {math.factorial(10):,}")

# Pattern 2: from module import specific names
from math import pi, sqrt, hypot

# Calculate the hypotenuse of a right triangle
leg_a = 3
leg_b = 4
hypotenuse = hypot(leg_a, leg_b)  # sqrt(a**2 + b**2)
print(f"\nHypotenuse of {leg_a}-{leg_b} triangle: {hypotenuse}")

# Circumference of a circle
radius = 7
circumference = 2 * pi * radius
print(f"Circumference of r={radius} circle: {circumference:.4f}")
```

```python
# Pattern 3: import with alias
import datetime as dt
import collections as col
import random as rnd

# datetime module — working with dates and times
today = dt.date.today()
now = dt.datetime.now()

print(f"Today:     {today}")
print(f"Now:       {now.strftime('%Y-%m-%d %H:%M:%S')}")
print(f"Year:      {today.year}")
print(f"Day name:  {today.strftime('%A')}")

# Date arithmetic
birthday = dt.date(2000, 6, 15)
age_days = (today - birthday).days
print(f"Days since 2000-06-15: {age_days:,}")

# random module — for simulations and sampling
rnd.seed(42)  # Seed for reproducibility
print(f"\nRandom int [1, 100]: {rnd.randint(1, 100)}")
print(f"Random float [0,1):  {rnd.random():.4f}")
print(f"Random choice:       {rnd.choice(['apple', 'banana', 'cherry'])}")

numbers = list(range(1, 11))
rnd.shuffle(numbers)
print(f"Shuffled 1-10: {numbers}")
print(f"Sample 3 from it: {rnd.sample(numbers, 3)}")
```

## Useful Standard Library Modules

Python's standard library is extensive. A few modules you will use constantly are `os` (operating system interactions), `sys` (Python runtime information), `collections` (specialised container types), and `datetime`. Knowing what the standard library provides prevents you from reinventing the wheel. The official documentation at docs.python.org/3/library/ is the authoritative reference.

```python
import os
import sys

# os module — file system and environment
print("=== os module ===")
print(f"Current directory: {os.getcwd()}")
print(f"Path separator:    {os.sep!r}")         # '/' on Unix, '\\' on Windows
print(f"Line separator:    {os.linesep!r}")    # OS-appropriate line ending
print(f"CPU count:         {os.cpu_count()}")

# Environment variables
path_var = os.environ.get("PATH", "not set")
print(f"PATH starts with:  {path_var[:50]}...")

# sys module — Python interpreter information  
print("\n=== sys module ===")
print(f"Python version:    {sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}")
print(f"Platform:          {sys.platform}")
print(f"Max integer:       {sys.maxsize:,}")
print(f"Default encoding:  {sys.getdefaultencoding()}")
print(f"Module search path (first 3): {sys.path[:3]}")
```

```python
from collections import Counter, defaultdict, namedtuple, deque

# Counter — count occurrences of elements
words = "the quick brown fox jumps over the lazy dog the fox".split()
word_count = Counter(words)
print("=== Counter ===")
print(word_count.most_common(3))  # 3 most frequent words
print(f"'the' appears: {word_count['the']} times")

# Counter supports set-like arithmetic
votes_a = Counter({"alice": 5, "bob": 3, "carol": 2})
votes_b = Counter({"alice": 2, "carol": 4, "david": 1})
print(f"Combined: {votes_a + votes_b}")

# defaultdict — dict with automatic default values for missing keys
print("\n=== defaultdict ===")
subject_grades = defaultdict(list)  # Default: empty list
records = [("Alice","Math",90),("Bob","Math",80),("Alice","Physics",88),("Bob","Physics",75)]
for student, subject, grade in records:
    subject_grades[subject].append((student, grade))
for subject, grades in subject_grades.items():
    print(f"  {subject}: {grades}")

# namedtuple — tuples with named fields (lightweight data class)
print("\n=== namedtuple ===")
Point3D = namedtuple('Point3D', ['x', 'y', 'z'])
origin = Point3D(0, 0, 0)
vector = Point3D(1.5, -2.3, 4.0)
print(f"Origin: {origin}")
print(f"Vector x={vector.x}, magnitude^2={vector.x**2 + vector.y**2 + vector.z**2:.2f}")

# deque — efficient append/pop from both ends
print("\n=== deque ===")
queue = deque(["first", "second", "third"], maxlen=5)
queue.appendleft("zeroth")  # Add to front — O(1)
queue.append("fourth")      # Add to back  — O(1)
print(queue)
```

## The `__name__ == '__main__'` Guard

When Python runs a file, it sets a special variable `__name__`. If you are running the file directly, `__name__` is set to `'__main__'`. If the file is being imported as a module, `__name__` is set to the module's name. The `if __name__ == '__main__':` guard lets you include code that should only run when the file is executed directly — like tests or a CLI interface — but not when it is imported.

```python
# Demonstrate __name__ in a Colab cell
print(f"In this notebook, __name__ is: {__name__!r}")
# Output: '__main__' — because we are running this directly

# When a module is imported, __name__ is the module's filename (without .py)
# This guard is standard practice for any script that is also a module

# Example structure of a well-organised Python script/module:
"""
# mymodule.py

def calculate_area(radius):
    import math
    return math.pi * radius ** 2

def main():
    # This only runs when executed directly: python mymodule.py
    r = 5
    print(f"Area of circle with radius {r}: {calculate_area(r):.2f}")

if __name__ == '__main__':
    main()          # Runs when: python mymodule.py
                    # Does NOT run when: import mymodule
"""
print("The pattern above prevents unwanted execution when importing.")
```

## Creating a Module with %%writefile

In Colab you can create Python module files using the `%%writefile` cell magic. This writes the cell content to a file on the Colab virtual machine's disk. You can then import it just like any other module. This demonstrates how real Python projects are organised — with code split across multiple files.

```python
%%writefile stats_utils.py
"""
stats_utils.py — Simple statistics utility module.
Demonstrates module structure for PyPath Lesson 11.
"""

def mean(data):
    """Return the arithmetic mean of a list of numbers."""
    if not data:
        raise ValueError("Cannot compute mean of empty sequence")
    return sum(data) / len(data)

def variance(data):
    """Return the population variance of a list of numbers."""
    if len(data) < 2:
        raise ValueError("Variance requires at least 2 data points")
    m = mean(data)
    return sum((x - m) ** 2 for x in data) / len(data)

def std_dev(data):
    """Return the population standard deviation."""
    return variance(data) ** 0.5

def median(data):
    """Return the median value."""
    if not data:
        raise ValueError("Cannot compute median of empty sequence")
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2
    return (sorted_data[mid - 1] + sorted_data[mid]) / 2 if n % 2 == 0 else sorted_data[mid]

def summary(data):
    """Return a dict with basic summary statistics."""
    return {
        "count":  len(data),
        "min":    min(data),
        "max":    max(data),
        "mean":   mean(data),
        "median": median(data),
        "std":    std_dev(data),
    }

if __name__ == '__main__':
    sample = [12, 18, 25, 9, 31, 14, 20, 17]
    print("Sample:", sample)
    for key, val in summary(sample).items():
        print(f"  {key:<8}: {val:.2f}")
```

```python
# Now import and use our module
import stats_utils

readings = [15.2, 18.7, 22.1, 14.8, 19.3, 21.5, 17.4, 16.9, 20.0, 18.1]

print("Sensor Readings Analysis")
print("=" * 35)
stats = stats_utils.summary(readings)
for stat, value in stats.items():
    print(f"  {stat:<8}: {value:.2f}")

# You can also import specific functions
from stats_utils import mean, std_dev
print(f"\nMean: {mean(readings):.2f}")
print(f"Std:  {std_dev(readings):.2f}")

# Inspect the module
print(f"\nModule docstring: {stats_utils.__doc__.strip()}")
print(f"Functions: {[name for name in dir(stats_utils) if not name.startswith('_')]}") 
```

## Installing Packages with pip and Virtual Environments

The Python Package Index (PyPI) hosts over 500,000 third-party packages. `pip` is the standard tool for installing them. In Colab, you install packages with `!pip install package_name`. In real projects, **virtual environments** (`venv`) create isolated Python installations so that packages installed for one project do not interfere with another.
