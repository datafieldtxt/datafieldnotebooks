## Basic Arithmetic

Python can be used as a powerful calculator. It supports all standard arithmetic operations. You can type mathematical expressions directly into a code cell and run them to see the result instantly. This is one of the simplest and most satisfying ways to start learning Python — just type a calculation and see the answer.

```python
# Basic arithmetic operations
print(10 + 3)   # Addition       -> 13
print(10 - 3)   # Subtraction    -> 7
print(10 * 3)   # Multiplication -> 30
print(10 / 3)   # Division       -> 3.3333... (always returns a float)
print(10 // 3)  # Floor division -> 3 (rounds down to nearest whole number)
print(10 % 3)   # Modulo         -> 1 (remainder after division)
print(10 ** 3)  # Exponentiation -> 1000 (10 to the power of 3)
```

```python
# Python respects standard mathematical order of operations (PEMDAS)
# Parentheses > Exponents > Multiply/Divide > Add/Subtract

result1 = 2 + 3 * 4      # Multiplication happens first: 2 + 12 = 14
result2 = (2 + 3) * 4    # Parentheses first: 5 * 4 = 20

print("Without parentheses:", result1)  # 14
print("With parentheses:   ", result2)  # 20
```

## Variables — Storing Information

A **variable** is a named container that stores a value in your program's memory. In Python, you create a variable simply by writing its name, an equals sign, and the value you want to store. There is no need to declare a type in advance — Python figures that out automatically. Variable names should be descriptive so your code is easy to read and understand.

```python
# Assigning values to variables
student_name = "Alice"
student_age = 22
course_score = 94.5
is_enrolled = True

# Using variables in print statements
print("Name:    ", student_name)
print("Age:     ", student_age)
print("Score:   ", course_score)
print("Enrolled:", is_enrolled)
```

```python
# Variables can be updated — the new value replaces the old one
counter = 0
print("Start:", counter)  # 0

counter = counter + 1     # Add 1 to counter and store the result
print("After +1:", counter)  # 1

counter += 1              # Shorthand for counter = counter + 1
print("After +=1:", counter)  # 2

# You can also do this with other operators: -=, *=, /=
counter *= 5
print("After *=5:", counter)  # 10
```

## Using print() Effectively

The `print()` function is your primary tool for displaying output. It can print text (called **strings**), numbers, or the value of any variable. You can print multiple values at once by separating them with commas — Python automatically adds a space between them. The `print()` function is essential for understanding what your program is doing at any point.

```python
# print() with multiple arguments — Python adds a space between them
first_name = "Maria"
last_name = "Garcia"
age = 28

print("Name:", first_name, last_name)
print("Age:", age)

# Using sep= to change the separator (default is a space)
print(first_name, last_name, sep="-")  # Maria-Garcia

# Using end= to change what prints at the end (default is a newline \n)
print("Loading", end="")
print("...done!")  # Prints on the same line: Loading...done!
```

```python
# f-strings (formatted string literals) — the modern way to embed variables in text
# Prefix the string with 'f' and put variables inside curly braces {}
product = "laptop"
price = 999.99
quantity = 3

print(f"Product: {product}")
print(f"Price: ${price}")
print(f"Total for {quantity} units: ${price * quantity}")
# Output: Total for 3 units: $2999.97
```

## Putting It All Together — A Mini Program

Let us combine everything from this lesson into a small, realistic program. This example calculates the total cost of an online order. Notice how variables make the code readable and how `print()` with f-strings creates nicely formatted output.

```python
# A simple order summary program

# Store information in variables
item_name = "Wireless Headphones"
unit_price = 79.99
quantity = 2
tax_rate = 0.08  # 8% tax

# Perform calculations
subtotal = unit_price * quantity
tax_amount = subtotal * tax_rate
total = subtotal + tax_amount

# Display a formatted receipt
print("===== Order Summary =====")
print(f"Item:     {item_name}")
print(f"Quantity: {quantity}")
print(f"Price:    ${unit_price:.2f} each")  # :.2f = 2 decimal places
print(f"Subtotal: ${subtotal:.2f}")
print(f"Tax:      ${tax_amount:.2f}")
print(f"Total:    ${total:.2f}")
print("========================")
```
