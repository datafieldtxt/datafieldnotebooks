## Arithmetic Operators

Python supports all standard arithmetic operations. The three operators that often surprise beginners are `//` (floor division, which rounds down to the nearest integer), `%` (modulo, which gives the remainder after division), and `**` (exponentiation). Floor division and modulo are related: `a == (a // b) * b + (a % b)` is always true. The modulo operator is particularly useful for checking divisibility, cycling through values, and working with time.

```python
# All arithmetic operators
print("Addition:        ", 17 + 5)   # 22
print("Subtraction:     ", 17 - 5)   # 12
print("Multiplication:  ", 17 * 5)   # 85
print("True division:   ", 17 / 5)   # 3.4  (always returns float)
print("Floor division:  ", 17 // 5)  # 3    (integer result, rounds toward -inf)
print("Modulo:          ", 17 % 5)   # 2    (remainder)
print("Exponentiation:  ", 2 ** 10)  # 1024
print("Negative power:  ", 2 ** -3)  # 0.125 (equals 1/8)

# Verify the relationship between //, %, and /
a, b = 17, 5
print(f"\nVerify: {a} == ({a}//{b})*{b} + {a}%{b} => {a == (a//b)*b + a%b}")
```

```python
# Practical uses of modulo (%)

# 1. Check if a number is even or odd
number = 47
if number % 2 == 0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")  # 47 is odd

# 2. Wrap around (cycling) — useful for clocks, circular buffers
hour = 25
wrapped_hour = hour % 24
print(f"Hour {hour} wraps to: {wrapped_hour}")  # 1

# 3. Convert total seconds to hours, minutes, seconds
total_seconds = 9045
hours   = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60
print(f"{total_seconds}s = {hours}h {minutes}m {seconds}s")  # 2h 30m 45s
```

## Comparison Operators

Comparison operators compare two values and return a boolean (`True` or `False`). Python supports six comparison operators: `==` (equal), `!=` (not equal), `<`, `>`, `<=`, and `>=`. One of Python's elegant features is **operator chaining** — you can write `1 < x < 10` instead of `1 < x and x < 10`, and Python evaluates it correctly. This makes range checks much more readable.

```python
# Comparison operators
price = 49.99
budget = 50.00

print(price == budget)   # False — exactly equal?
print(price != budget)   # True  — not equal?
print(price < budget)    # True  — less than?
print(price > budget)    # False — greater than?
print(price <= budget)   # True  — less than or equal?
print(price >= budget)   # False — greater than or equal?

# String comparison (lexicographic — based on character codes)
print("apple" < "banana")  # True  — 'a' comes before 'b'
print("abc" == "abc")      # True
print("ABC" == "abc")      # False — case matters!
```

```python
# Operator chaining — Python's elegant range checks
temperature = 22

# Standard way (works in most languages)
if temperature > 18 and temperature < 30:
    print("Comfortable temperature (standard)")

# Python's chained comparison — reads like math!
if 18 < temperature < 30:
    print("Comfortable temperature (chained)")  # This prints

# Chaining works with any comparisons
x = 5
print(1 < x < 10)    # True
print(1 < x < 4)     # False
print(0 <= x <= 5)   # True
print(1 == 1 == 1)   # True

# Warning: chaining is evaluated left to right with AND logic
print(1 < 3 > 2)  # True  — means (1 < 3) and (3 > 2)
```

## Logical Operators

Logical operators (`and`, `or`, `not`) combine boolean expressions. Python uses **short-circuit evaluation**: with `and`, if the left side is `False`, the right side is never evaluated; with `or`, if the left side is `True`, the right side is skipped. This behaviour is not just an optimization — it is frequently used to avoid errors (e.g., `if lst and lst[0] == x:` avoids an IndexError on an empty list).

```python
# Logical operators: and, or, not
is_adult = True
has_ticket = False
is_vip = True

# and — True only if BOTH sides are True
print(is_adult and has_ticket)   # False
print(is_adult and is_vip)       # True

# or — True if AT LEAST ONE side is True
print(has_ticket or is_vip)      # True
print(has_ticket or False)       # False

# not — inverts the boolean value
print(not has_ticket)            # True
print(not is_adult)              # False

# Combining logical operators (use parentheses to clarify intent)
can_enter = is_adult and (has_ticket or is_vip)
print(f"Can enter: {can_enter}")  # True
```

```python
# Truth table — all combinations of and/or
print("A      B      A and B  A or B")
print("-" * 36)
for a in [True, False]:
    for b in [True, False]:
        print(f"{str(a):<7}{str(b):<7}{str(a and b):<9}{str(a or b)}")

# Short-circuit evaluation in practice
users = ["alice", "bob"]

# Safe: 'and' short-circuits — if list is empty, index is never checked
if users and users[0] == "alice":
    print("\nFirst user is alice")  # Safe even if 'users' were empty

# 'or' for default values — a common Python pattern
config_port = None
port = config_port or 8080  # If config_port is falsy, use 8080
print(f"Using port: {port}")  # 8080
```

## Bitwise Operators

Bitwise operators work directly on the binary (bit-level) representation of integers. While they appear in lower-level code, they are practically useful for working with **flags** (storing multiple on/off states in a single integer), **masks** (selecting specific bits), and certain performance-sensitive algorithms. Understanding them also deepens your understanding of how computers represent data.

```python
# Bitwise operators work on the binary representation of integers
a = 0b1010   # 10 in decimal (binary: 1010)
b = 0b1100   # 12 in decimal (binary: 1100)

print(f"a = {a} (binary: {bin(a)})")
print(f"b = {b} (binary: {bin(b)})")
print()

# AND (&) — bit is 1 only if BOTH bits are 1
print(f"a & b  = {a & b}  (binary: {bin(a & b)})")   # 8  (1000)

# OR (|) — bit is 1 if EITHER bit is 1
print(f"a | b  = {a | b}  (binary: {bin(a | b)})")   # 14 (1110)

# XOR (^) — bit is 1 if bits are DIFFERENT
print(f"a ^ b  = {a ^ b}   (binary: {bin(a ^ b)})")  # 6  (0110)

# NOT (~) — flips all bits (result is -(n+1) for positive integers)
print(f"~a     = {~a}  (flips all bits)")              # -11

# Left shift (<<) — multiply by 2 for each shift
print(f"a << 1 = {a << 1}  (binary: {bin(a << 1)})")  # 20 (multiply by 2)

# Right shift (>>) — divide by 2 for each shift (integer division)
print(f"a >> 1 = {a >> 1}   (binary: {bin(a >> 1)})") # 5  (divide by 2)
```

```python
# Practical use: Permission flags (like Unix file permissions)
# Store multiple boolean states in a single integer using bits as flags

READ    = 0b001  # 1 — bit 0
WRITE   = 0b010  # 2 — bit 1
EXECUTE = 0b100  # 4 — bit 2

# Grant read and write permissions
user_permissions = READ | WRITE
print(f"Permissions: {bin(user_permissions)} = {user_permissions}")

# Check if a specific permission is set using AND
has_read    = bool(user_permissions & READ)
has_write   = bool(user_permissions & WRITE)
has_execute = bool(user_permissions & EXECUTE)
print(f"Read: {has_read}, Write: {has_write}, Execute: {has_execute}")
# Read: True, Write: True, Execute: False

# Revoke write permission using AND with NOT
user_permissions &= ~WRITE
print(f"After revoking WRITE: {bin(user_permissions)} = {user_permissions}")
print(f"Write now: {bool(user_permissions & WRITE)}")  # False
```

## Operator Precedence

When an expression contains multiple operators, Python uses a strict **precedence order** to decide which operations to perform first. This mirrors mathematical conventions: exponentiation binds most tightly, then multiplication/division, then addition/subtraction, then comparisons, then logical operators. When in doubt — or when clarity matters — use parentheses. Explicit parentheses never hurt and often make code significantly easier to understand.

```python
# Operator precedence — from highest to lowest (simplified)
# 1. ()              Parentheses
# 2. **              Exponentiation
# 3. +x, -x, ~x     Unary operators
# 4. *, /, //, %     Multiplication, division
# 5. +, -            Addition, subtraction
# 6. <<, >>          Bitwise shifts
# 7. &               Bitwise AND
# 8. ^               Bitwise XOR
# 9. |               Bitwise OR
# 10. ==, !=, <, >   Comparisons
# 11. not            Logical NOT
# 12. and            Logical AND
# 13. or             Logical OR

# Examples showing precedence in action
print(2 + 3 * 4)       # 14  — multiply first, then add
print((2 + 3) * 4)     # 20  — parentheses override precedence
print(2 ** 3 ** 2)     # 512 — ** is RIGHT-associative: 2**(3**2) = 2**9 = 512
print((2 ** 3) ** 2)   # 64  — left-to-right: 8**2 = 64
print(-2 ** 2)         # -4  — unary minus applies AFTER **: -(2**2)
print((-2) ** 2)       # 4   — parentheses make it (-2)**2 = 4

# Mixed arithmetic and comparison
print(1 + 2 == 3)      # True  — arithmetic first (1+2=3), then compare
print(True and 2 > 1)  # True  — comparison first, then and
```
