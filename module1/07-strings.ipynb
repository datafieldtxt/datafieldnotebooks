## String Slicing

Because strings are sequences, you can access individual characters by index (`s[0]`) or extract substrings using slicing (`s[start:stop:step]`). Python uses **zero-based indexing** — the first character is at index 0. Negative indices count from the end: `-1` is the last character. Slicing never raises an IndexError even if your indices are out of range. Understanding slicing is essential for string parsing and manipulation.

```python
# String indexing and slicing
text = "Hello, Python!"

# Individual character access
print(text[0])    # H  — first character
print(text[-1])   # !  — last character
print(text[-7])   # P  — 7th from end

# Slicing: [start:stop]  — stop is exclusive
print(text[0:5])   # Hello  — characters 0,1,2,3,4
print(text[7:13])  # Python — characters 7 through 12
print(text[:5])    # Hello  — omit start = from beginning
print(text[7:])    # Python! — omit stop = to the end
print(text[:])     # Hello, Python! — full copy

# Slicing with step: [start:stop:step]
print(text[::2])   # Hlo yhn — every 2nd character
print(text[::-1])  # !nohtyP ,olleH — reversed string (step=-1)

# Practical: extract the domain from an email
email = "alice.smith@example.com"
at_index = email.index("@")
username = email[:at_index]
domain = email[at_index + 1:]
print(f"Username: {username}, Domain: {domain}")
```

## Common String Methods

Python strings come with a rich set of built-in methods. Because strings are immutable, every method returns a **new** string — the original is unchanged. The most commonly used methods cover case conversion, whitespace handling, searching, splitting, and joining. Knowing these methods lets you manipulate text without writing loops for most common tasks.

```python
# Case methods
sentence = "  The Quick Brown Fox  "
print(sentence.upper())       # '  THE QUICK BROWN FOX  '
print(sentence.lower())       # '  the quick brown fox  '
print(sentence.title())       # '  The Quick Brown Fox  ' (each word capitalised)
print(sentence.swapcase())    # '  tHE qUICK bROWN fOX  '

# Whitespace removal
print(repr(sentence.strip()))   # 'The Quick Brown Fox'
print(repr(sentence.lstrip()))  # 'The Quick Brown Fox  '
print(repr(sentence.rstrip())) # '  The Quick Brown Fox'

# strip() can also remove specific characters
filepath = "///home/user/docs///"
print(filepath.strip("/"))  # 'home/user/docs'
```

```python
# Search, replace, split, join
review = "The product quality is good but the price is too high for the quality"

# find() returns index of first occurrence, or -1 if not found
print(review.find("quality"))     # 12
print(review.find("quality", 20)) # 65 (search starting at index 20)
print(review.find("delivery"))    # -1 (not found)

# count() counts non-overlapping occurrences
print(review.count("quality"))    # 2
print(review.count("the"))        # 2 (case-sensitive)

# replace() — returns new string
cleaned = review.replace("too high", "reasonable")
print(cleaned)

# split() — splits into a list of strings
words = review.split()            # Split on any whitespace by default
print(f"Word count: {len(words)}")  # 15

csv_line = "Alice,28,Engineer,Toronto"
fields = csv_line.split(",")
print(fields)  # ['Alice', '28', 'Engineer', 'Toronto']

# join() — the inverse of split()
tags = ["python", "beginner", "programming"]
print(" | ".join(tags))       # python | beginner | programming
print("-".join("HELLO"))      # H-E-L-L-O (join characters)

# Check methods
print("hello123".isalnum())  # True
print("  ".isspace())        # True
print("hello".startswith("hel"))  # True
print("report.pdf".endswith(".pdf"))  # True
```

## f-Strings and Format Specifications

f-strings (formatted string literals, introduced in Python 3.6) are the modern, preferred way to embed expressions in strings. Prefix the string with `f` and place any expression inside `{}`. f-strings support a powerful **format specification mini-language** inside the braces after a colon, which controls alignment, padding, number formatting, and more. They are faster and more readable than older formatting methods.

```python
# f-string basics
product = "Laptop"
price = 1299.5
stock = 42

print(f"Product: {product}")
print(f"Price: ${price}")
print(f"In stock: {stock} units")
print(f"Total value: ${price * stock:,.2f}")  # Expression inside {}

# Format specifications: {value:format_spec}
# Number formatting
print(f"{price:.2f}")         # 1299.50  — 2 decimal places
print(f"{price:10.2f}")       # '   1299.50' — width 10, 2 decimal places
print(f"{price:0>10.2f}")     # '001299.50' — zero-padded
print(f"{1234567.89:,.2f}")   # 1,234,567.89 — thousands separator
print(f"{0.456789:.1%}")      # 45.7%  — percentage format
print(f"{255:08b}")           # 11111111 — binary, zero-padded to 8 digits
print(f"{255:#x}")            # 0xff — hexadecimal with prefix
```

```python
# Alignment and padding in f-strings
header = f"{'Name':<15} {'Score':>6} {'Grade':^8}"
print(header)
print("-" * len(header))

students = [("Alice Johnson", 94, "A"), ("Bob", 78, "C"), ("Carol Martinez", 88, "B")]
for name, score, grade in students:
    print(f"{name:<15} {score:>6} {grade:^8}")

# Debugging with f-strings — use = to show variable name and value
x = 42
y = [1, 2, 3]
print(f"\n{x=}")    # x=42  (Python 3.8+)
print(f"{y=}")     # y=[1, 2, 3]
print(f"{x*2+1=}") # x*2+1=85
```

## str.format() and Raw Strings

`str.format()` is the older formatting method that predates f-strings but is still widely used, especially when the format string is stored in a variable or read from a configuration file. **Raw strings** (prefixed with `r`) treat backslashes as literal characters — they are essential for Windows file paths and regular expression patterns where backslashes have special meaning.

```python
# str.format() — positional and keyword placeholders
template = "Dear {name},\n\nYour order #{order_id} totalling ${total:.2f} has shipped."
print(template.format(name="Alice", order_id=4521, total=87.50))

# Positional: {0}, {1}, etc.
print("{0} and {1} are friends. {0} likes Python.".format("Alice", "Bob"))

# Reusing a value multiple times
print("{0} * {0} = {1}".format(7, 49))

# Raw strings — backslash is NOT an escape character
windows_path = r"C:\Users\Alice\Documents\report.txt"
print(windows_path)    # C:\Users\Alice\Documents\report.txt
print(len(windows_path))  # Backslashes are literal characters

# Without r prefix, \U and \D would be invalid escape sequences
normal_path = "C:\\Users\\Alice"  # Must double-escape backslashes
print(normal_path)

# Raw strings in regular expressions
import re
# Without r, \d would need to be \\d
digit_pattern = r"\d+"  # Matches one or more digits
print(re.findall(digit_pattern, "Order 42 costs $19.99"))  # ['42', '19', '99']
```

## Introduction to Regular Expressions

The `re` module provides powerful tools for matching patterns in strings. A **regular expression** (regex) is a pattern-matching language. It lets you find, extract, and replace text that matches a pattern — things that would be complex loops become single function calls. The most important functions are `re.search()` (find first match anywhere), `re.findall()` (find all matches), and `re.sub()` (find and replace).

```python
import re

# re.search() — find the first match anywhere in the string
log_line = "2024-01-15 ERROR: Database connection failed on attempt 3"

# Search for a date pattern (YYYY-MM-DD)
date_match = re.search(r"\d{4}-\d{2}-\d{2}", log_line)
if date_match:
    print(f"Date found: {date_match.group()}")  # 2024-01-15

# re.findall() — return all non-overlapping matches as a list
numbers = re.findall(r"\d+", log_line)
print(f"Numbers in log line: {numbers}")  # ['2024', '01', '15', '3']

# re.sub() — replace matches with a new string
masked = re.sub(r"\d{4}-\d{2}-\d{2}", "[DATE]", log_line)
print(f"Masked: {masked}")

# re.split() — split on a pattern
text = "one  two   three    four"  # Variable whitespace
words = re.split(r"\s+", text)  # Split on one or more whitespace chars
print(f"Words: {words}")
```

```python
import re

# Practical: extract emails from a block of text
text = """
Contact our team:
- Sales: sales@company.com
- Support: help.desk@support.co.uk
- Admin: admin_2024@internal.org
"""

# Email pattern: word chars/dots before @, then domain, then TLD
email_pattern = r"[\w.+-]+@[\w-]+\.[\w.]+"
emails = re.findall(email_pattern, text)
print("Emails found:")
for email in emails:
    print(f"  {email}")

# Practical: validate a simple phone number format
def is_valid_phone(phone):
    pattern = r"^\+?1?[-.\s]?\(?\d{3}\)?[-.\s]?\d{3}[-.\s]?\d{4}$"
    return bool(re.match(pattern, phone.strip()))

test_phones = ["+1-416-555-1234", "416.555.1234", "(416) 555-1234", "not-a-phone", "123"]
for phone in test_phones:
    status = "valid" if is_valid_phone(phone) else "invalid"
    print(f"{phone:<20} -> {status}")
```
