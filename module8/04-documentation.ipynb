## Docstring Formats — Three Styles for the Same Function

Python supports three widely-used docstring formats. Choose one and apply it consistently throughout your project.

````python
# ── GOOGLE STYLE ─────────────────────────────────────────────────────────────
# Most readable, widely used with VS Code and Sphinx napoleon extension

def calculate_bmi_google(weight_kg: float, height_m: float) -> float:
    """Calculate Body Mass Index (BMI) for a person.

    BMI is defined as weight divided by the square of height.
    The result is categorised as: underweight (<18.5), normal (18.5-24.9),
    overweight (25-29.9), or obese (>=30).

    Args:
        weight_kg: Body weight in kilograms. Must be positive.
        height_m: Height in metres. Must be positive and non-zero.

    Returns:
        The BMI value rounded to one decimal place.

    Raises:
        ValueError: If weight_kg or height_m is not positive.
        ZeroDivisionError: If height_m is zero.

    Example:
        >>> calculate_bmi_google(70, 1.75)
        22.9
        >>> calculate_bmi_google(90, 1.80)
        27.8
    """
    if weight_kg <= 0 or height_m <= 0:
        raise ValueError("Weight and height must be positive values.")
    return round(weight_kg / height_m ** 2, 1)

print(calculate_bmi_google(70, 1.75))
print(calculate_bmi_google.__doc__[:300])
````

````python
# ── NUMPY STYLE ──────────────────────────────────────────────────────────────
# Popular in scientific computing (NumPy, SciPy, pandas)

def calculate_bmi_numpy(weight_kg: float, height_m: float) -> float:
    """
    Calculate Body Mass Index (BMI) for a person.

    BMI is defined as weight divided by the square of height.

    Parameters
    ----------
    weight_kg : float
        Body weight in kilograms. Must be positive.
    height_m : float
        Height in metres. Must be positive and non-zero.

    Returns
    -------
    float
        The BMI value rounded to one decimal place.

    Raises
    ------
    ValueError
        If weight_kg or height_m is not positive.

    Examples
    --------
    >>> calculate_bmi_numpy(70, 1.75)
    22.9
    >>> calculate_bmi_numpy(90, 1.80)
    27.8

    Notes
    -----
    BMI categories (WHO standard):
    - Underweight: < 18.5
    - Normal weight: 18.5 - 24.9
    - Overweight: 25.0 - 29.9
    - Obese: >= 30.0
    """
    if weight_kg <= 0 or height_m <= 0:
        raise ValueError("Weight and height must be positive values.")
    return round(weight_kg / height_m ** 2, 1)

print("NumPy style docstring written.")
print(calculate_bmi_numpy(70, 1.75))
````

````python
# ── reStructuredText (reST) STYLE ─────────────────────────────────────────────
# Classic Sphinx style — verbose but widely supported

def calculate_bmi_rst(weight_kg: float, height_m: float) -> float:
    """
    Calculate Body Mass Index (BMI) for a person.

    BMI is defined as weight divided by the square of height.

    :param weight_kg: Body weight in kilograms. Must be positive.
    :type weight_kg: float
    :param height_m: Height in metres. Must be positive and non-zero.
    :type height_m: float
    :returns: The BMI value rounded to one decimal place.
    :rtype: float
    :raises ValueError: If weight_kg or height_m is not positive.

    Usage example::

        >>> calculate_bmi_rst(70, 1.75)
        22.9
    """
    if weight_kg <= 0 or height_m <= 0:
        raise ValueError("Weight and height must be positive values.")
    return round(weight_kg / height_m ** 2, 1)

print("reST style docstring written.")
print(calculate_bmi_rst(90, 1.80))
````

## Doctests — Executable Examples in Docstrings

Doctest examples (lines starting with `>>>`) can be run automatically as tests. This keeps documentation and implementation in sync.

````python
%%writefile geometry.py
"""Geometry utility functions."""
import math


def circle_area(radius: float) -> float:
    """Return the area of a circle.

    Args:
        radius: The radius of the circle. Must be non-negative.

    Returns:
        Area of the circle (pi * r^2).

    Raises:
        ValueError: If radius is negative.

    Examples:
        >>> circle_area(0)
        0.0
        >>> circle_area(1)
        3.141592653589793
        >>> round(circle_area(5), 4)
        78.5398
        >>> circle_area(-1)
        Traceback (most recent call last):
            ...
        ValueError: Radius must be non-negative.
    """
    if radius < 0:
        raise ValueError("Radius must be non-negative.")
    return math.pi * radius ** 2


def rectangle_perimeter(width: float, height: float) -> float:
    """Return the perimeter of a rectangle.

    Args:
        width: Width of the rectangle.
        height: Height of the rectangle.

    Returns:
        Perimeter (2 * (width + height)).

    Examples:
        >>> rectangle_perimeter(4, 6)
        20
        >>> rectangle_perimeter(3.5, 2.5)
        12.0
    """
    return 2 * (width + height)


if __name__ == "__main__":
    import doctest
    doctest.testmod(verbose=True)
````

## Sphinx — Generate HTML Documentation

Sphinx is the standard tool for generating Python project documentation. It reads your docstrings and creates beautiful HTML, PDF, and ePub output.

````python
import os

# Create a minimal Sphinx docs directory structure
os.makedirs("/tmp/sphinx_demo/docs", exist_ok=True)
os.makedirs("/tmp/sphinx_demo/mylib", exist_ok=True)

# Write a sample library module
with open("/tmp/sphinx_demo/mylib/__init__.py", "w") as f:
    f.write('"""My Library."""\nfrom .core import greet, add\n')

with open("/tmp/sphinx_demo/mylib/core.py", "w") as f:
    f.write('''"""Core functions for mylib."""\n\n\ndef greet(name: str) -> str:\n    """Return a personalised greeting.\n\n    Args:\n        name: The name to greet.\n\n    Returns:\n        A greeting string.\n\n    Examples:\n        >>> greet("World")\n        \'Hello, World!\'\n    """\n    return f"Hello, {name}!"\n\n\ndef add(a: float, b: float) -> float:\n    """Add two numbers.\n\n    Args:\n        a: First operand.\n        b: Second operand.\n\n    Returns:\n        The sum of a and b.\n\n    Examples:\n        >>> add(2, 3)\n        5\n    """\n    return a + b\n''')

print("Sample library created at /tmp/sphinx_demo/mylib/")
````

````python
%%writefile /tmp/sphinx_demo/docs/conf.py
import os
import sys

# Add the project root to sys.path so autodoc can find the modules
sys.path.insert(0, os.path.abspath(".."))

# ── Project information ───────────────────────────────────────────────────────
project = "MyLib"
copyright = "2024, Ada Lovelace"
author = "Ada Lovelace"
release = "0.1.0"

# ── Extensions ────────────────────────────────────────────────────────────────
extensions = [
    "sphinx.ext.autodoc",        # read docstrings automatically
    "sphinx.ext.napoleon",       # support Google & NumPy docstring styles
    "sphinx.ext.viewcode",       # add [source] links to rendered docs
    "sphinx.ext.doctest",        # run doctest examples in docs
    "sphinx.ext.intersphinx",    # link to other projects' docs
]

# ── Napoleon settings (Google/NumPy docstrings) ───────────────────────────────
napoleon_google_docstring = True
napoleon_numpy_docstring = True
napoleon_include_init_with_doc = True

# ── Intersphinx mappings ──────────────────────────────────────────────────────
intersphinx_mapping = {
    "python": ("https://docs.python.org/3", None),
    "numpy": ("https://numpy.org/doc/stable", None),
}

# ── HTML output ───────────────────────────────────────────────────────────────
html_theme = "sphinx_rtd_theme"   # pip install sphinx-rtd-theme
html_static_path = ["_static"]
````

## MkDocs + mkdocstrings — Simpler Alternative

MkDocs uses Markdown instead of reStructuredText, making it easier for most developers.

````python
%%writefile /tmp/sphinx_demo/mkdocs.yml
site_name: MyLib Documentation
site_description: Documentation for MyLib
site_author: Ada Lovelace
repo_url: https://github.com/ada/mylib

theme:
  name: material            # pip install mkdocs-material
  features:
    - navigation.tabs
    - search.suggest

plugins:
  - search
  - mkdocstrings:
      handlers:
        python:
          options:
            docstring_style: google   # or numpy
            show_source: true
            show_root_heading: true

nav:
  - Home: index.md
  - API Reference:
    - Core: api/core.md
````

````python
# mkdocstrings API page uses ::: directives
import os
os.makedirs("/tmp/sphinx_demo/mkdocs_docs/api", exist_ok=True)

with open("/tmp/sphinx_demo/mkdocs_docs/api/core.md", "w") as f:
    f.write("# Core Module\n\n::: mylib.core\n")

with open("/tmp/sphinx_demo/mkdocs_docs/index.md", "w") as f:
    f.write("# MyLib\n\nWelcome to the MyLib documentation.\n")

print("MkDocs structure created.")
print("\nTo build: mkdocs build")
print("To serve locally: mkdocs serve  (at http://127.0.0.1:8000)")
````

## Type Hints + Docstrings = Full API Documentation

Combining type hints (for tools and IDEs) with docstrings (for humans) gives you the best of both worlds.

````python
from __future__ import annotations
from typing import Sequence
from dataclasses import dataclass


@dataclass
class DataPoint:
    """A single measurement data point.

    Attributes:
        timestamp: Unix timestamp of the measurement.
        value: The measured value.
        label: Optional category label.
    """
    timestamp: float
    value: float
    label: str | None = None


def normalise(
    data: Sequence[DataPoint],
    *,
    method: str = "minmax",
) -> list[DataPoint]:
    """Normalise a sequence of data points.

    Scales the ``value`` field of each data point to the [0, 1] range
    using the specified normalisation method.

    Args:
        data: The sequence of data points to normalise. Must be non-empty.
        method: Normalisation method. One of ``'minmax'`` (default) or
            ``'zscore'``. Min-max scaling maps values to [0, 1];
            z-score standardisation uses mean and standard deviation.

    Returns:
        A new list of :class:`DataPoint` objects with normalised values.
        Original objects are not modified.

    Raises:
        ValueError: If ``data`` is empty or ``method`` is not recognised.

    Example:
        >>> pts = [DataPoint(1.0, 0.0), DataPoint(2.0, 5.0), DataPoint(3.0, 10.0)]
        >>> normalised = normalise(pts)
        >>> [round(p.value, 2) for p in normalised]
        [0.0, 0.5, 1.0]
    """
    if not data:
        raise ValueError("data must not be empty")
    if method == "minmax":
        min_v = min(p.value for p in data)
        max_v = max(p.value for p in data)
        spread = max_v - min_v or 1.0
        return [DataPoint(p.timestamp, (p.value - min_v) / spread, p.label) for p in data]
    elif method == "zscore":
        n = len(data)
        mean = sum(p.value for p in data) / n
        variance = sum((p.value - mean) ** 2 for p in data) / n
        std = variance ** 0.5 or 1.0
        return [DataPoint(p.timestamp, (p.value - mean) / std, p.label) for p in data]
    else:
        raise ValueError(f"Unknown method: {method!r}. Choose 'minmax' or 'zscore'.")


# Demo
points = [DataPoint(1.0, 0.0), DataPoint(2.0, 5.0), DataPoint(3.0, 10.0)]
normalised = normalise(points)
for p in normalised:
    print(f"  t={p.timestamp}  value={p.value:.3f}")
````
